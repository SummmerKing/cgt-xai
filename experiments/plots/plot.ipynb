```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Feature_Name': ['Diet Score', 'Sleep Duration', 'Family History', 'BMI'],
    'Feature_Label': ['diet_score', 'sleep_duration', 'family_history', 'bmi'],
    'SHAP': [2.898, 2.787, 2.960, 2.528],
    'Causal SHAP': [0.037, -0.031, 0.001, -0.240],
    'Least Core': [2.886, 2.681, 3.109, 2.498],
    'Nucleolus': [2.886, 2.681, 3.109, 2.498],
    'Gately': [2.886, 2.837, 2.903, 2.547]
}

df = pd.DataFrame(data)

# 2. Setup the plot
plt.figure(figsize=(8, 6))

# 3. Create the bar chart
# We use 'Feature_Label' for X (to match the image's snake_case)
# and 'Causal SHAP' for Y (to match the image's values)
plt.bar(df['Feature_Label'], df['Causal SHAP'], width=0.5, color='#1f77b4')

# 4. Formatting to match the image
plt.title('Feature Attribution — Causal SHAP', fontsize=12, fontweight='bold')
plt.ylabel('Attribution')

# Add the horizontal grid lines
plt.grid(axis='y', linestyle='-', alpha=0.4)
plt.axhline(0, color='lightgrey', linewidth=1) # Explicit line at 0

# Rotate x-axis labels to match the slant
plt.xticks(rotation=45, ha='right')

# Adjust layout to prevent clipping of labels
plt.tight_layout()

# 5. Display the graph
plt.show()
```
